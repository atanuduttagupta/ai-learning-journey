## 1. What is KNN?

KNN (K-Nearest Neighbors) is a **non-parametric, lazy learning algorithm**.

### Key Characteristics:
- **Non-parametric** → does not assume any data distribution
- **Lazy learner** → no training phase
- **Instance-based** → stores entire dataset

### How it works:
1. Choose K (number of neighbors)
2. Compute distance from query point to all points
3. Select K nearest points
4. Predict:
   - Classification → majority vote
   - Regression → average

### Real-world intuition:
👉 "Tell me who your closest neighbors are, and I will tell you who you are."



## 2. Distance Metrics (DETAILED)

Distance defines "similarity" in KNN.

---

### 2.1 Euclidean Distance

Formula:

$$
d = \sqrt{\sum (x_i - y_i)^2}
$$

👉 Straight-line distance between two points.

**Use case:**
- Continuous numerical features
- When geometry matters

---

### 2.2 Manhattan Distance

Formula:

$$
d = \sum |x_i - y_i|
$$

👉 Distance along grid paths (like city blocks).

**Use case:**
- When movement is restricted (e.g., grid-based systems)
- Robust to outliers

---

### 2.3 Minkowski Distance

Formula:

$$
d = \left( \sum |x_i - y_i|^p \right)^{1/p}
$$

- p = 1 → Manhattan
- p = 2 → Euclidean

👉 Generalized distance metric

---

### 2.4 Cosine Similarity

Formula:

$$
cos(\theta) = \frac{A \cdot B}{||A|| ||B||}
$$

👉 Measures angle between vectors (not magnitude)

**Use case:**
- Text similarity
- High-dimensional embeddings (NLP, search)

---


🏭 When to use what (VERY IMPORTANT)

### Use Euclidean when:

Features are continuous

Geometry matters (image, physics, clustering)

### Use Manhattan when:

Data has outliers

Movement is constrained (grid systems)

High-dimensional sparse data


## 3. Choosing K (DETAILED)

K controls model complexity.

### Small K:
- Very flexible
- Captures noise → **overfitting**

### Large K:
- Smooth decision boundary
- Misses patterns → **underfitting**

### Balanced K:
- Optimal bias-variance tradeoff

### Rule of thumb:
$$
K \approx \sqrt{N}
$$

### Use case:
- Tune K using validation or cross-validation



## 4. Feature Scaling (CRITICAL)

Distance is sensitive to scale.

Example:
- Salary: 1,00,000
- Age: 25

👉 Salary dominates distance calculation

### Solution:
- Standardization (Z-score)
- Normalization (Min-Max)

---


🧠 What it does

Centers data around mean = 0

Scales data to standard deviation = 1

📊 Output range

Not fixed (can be -3, +2, etc.)
📍 Example

Original: [1000, 2000, 3000]
After standardization: [-1.22, 0, 1.22]

🎯 When to use

When data follows normal (Gaussian) distribution

Algorithms that assume centered data:

Logistic Regression

SVM

KNN (very common)

```python

from sklearn.preprocessing import StandardScaler
import numpy as np

X = np.array([[1000,1],[2000,2],[3000,3]])
scaler = StandardScaler()
scaler.fit_transform(X)
```

Normalization (Min-Max Scaling)

x_scaled = (x - x_min)/(x_max - x_min)


🧠 What it does

Scales data to a fixed range [0, 1]

📊 Output range

Always between 0 and 1

📍 Example

Original: [1000, 2000, 3000]

After normalization: [0, 0.5, 1]

🎯 When to use

When scale matters strictly

Neural Networks (especially)

Image data (pixel values)

Standardization → “center and normalize spread”

Normalization → “compress everything into same box

Large Outlier - Use Z Score Standardization

```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np

X = np.array([[1000,1],
              [2000,2],
              [3000,3]])

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

print(X_scaled)
```


## 5. Decision Boundary

KNN creates **non-linear boundaries**.

### Insight:
- K = 1 → highly irregular boundary
- K large → smoother boundary

### Use case:
- Complex pattern recognition (image classification)



## 6. Classification vs Regression

### Classification:
- Output: class label
- Method: majority voting

### Regression:
- Output: continuous value
- Method: average of neighbors



## 7. Weighted KNN

Not all neighbors are equally important.

### Idea:
- Closer neighbors → higher weight

Formula:

$$
Weight = \frac{1}{distance}
$$

### Use case:
- When closer points are more reliable


```python
from sklearn.neighbors import KNeighborsClassifier
import numpy as np

# Simple dataset
X = np.array([[1], [2], [3], [10]])
y = np.array([0, 0, 1, 1])

# Query point
query = [[2.5]]

# 🔹 Normal KNN
knn_normal = KNeighborsClassifier(n_neighbors=3)
knn_normal.fit(X, y)
print("Normal KNN Prediction:", knn_normal.predict(query))

# 🔹 Weighted KNN
knn_weighted = KNeighborsClassifier(n_neighbors=3, weights='distance')
knn_weighted.fit(X, y)
print("Weighted KNN Prediction:", knn_weighted.predict(query))
```


## 8. Curse of Dimensionality

As dimensions increase:
- Distance differences shrink
- All points become equally distant

👉 KNN loses effectiveness

### Use case:
- High-dimensional data (NLP, embeddings)
- Requires dimensionality reduction (PCA)



## 9. Computational Complexity

- Training: O(1)
- Prediction: O(N × D)

Where:
- N = number of samples
- D = number of features

👉 Slow for large datasets



## 10. Optimizations

### KD Tree:
- Efficient for low dimensions

### Ball Tree:
- Better for higher dimensions

### ANN (Approximate Nearest Neighbors):
- Faster, approximate results

### Use case:
- Large-scale systems (recommendation engines)

- These reduce search from:

O(N) → ~O(log N) (in ideal cases)



## 11. Limitations



- Slow inference
- Sensitive to noise
- Requires scaling
- Poor in high dimensions
