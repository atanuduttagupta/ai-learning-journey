# 🔹 1. What is a Loss Function?

In Machine Learning, we make predictions.

👉 But predictions are rarely perfect.

So we need a way to measure **how wrong we are**.

That measure is called a **Loss Function**.

### 🧠 Example:
If actual value = 10 and prediction = 8
Error = 2

We square it to penalize larger errors:
Loss = (10 - 8)^2 = 4

👉 Bigger mistake → much bigger penalty

```python
# Simple loss example
y_true = 10
y_pred = 8

loss = (y_true - y_pred)**2
print('Loss:', loss)
```

# 🔹 2. Mean Squared Error (MSE)

In real problems, we have multiple data points.

So we take the **average loss**:

MSE = mean of squared errors

### Why squared?
- Removes negative sign
- Penalizes large errors more


```python
import numpy as np

y_true = np.array([2,4,6])
y_pred = np.array([1,5,7])

mse = np.mean((y_true - y_pred)**2)
print('MSE:', mse)
```

# 🔹 3. What is a Gradient?

Gradient = **rate of change of loss**

👉 It tells us:
- Which direction to move
- How steep the slope is

### 🧠 Intuition:
Think of a mountain:
- Gradient tells you slope direction
- You want to go DOWN (minimum loss)


# 🔹 4. Gradient Descent Intuition

We update parameters using:

theta = theta - learning_rate * gradient

### Meaning:
- Gradient → direction
- Learning rate → step size

👉 We move step-by-step towards minimum loss

# 🔹 5. Learning Rate (Critical Concept)

Learning rate controls **how fast we learn**.

### Cases:
- Too small → very slow learning
- Too large → overshoot / diverge
- Just right → smooth convergence


# 🔹 6. Gradient Descent Implementation

We will train a simple model:
y = w * x

Goal: Learn w using gradient descent


```python
import numpy as np

# Data (true relation y = 2x)
X = np.array([1,2,3,4])
y = np.array([2,5,7,9])

# Initialize weight
w = 0
lr = 0.01

for epoch in range(100):
    # Prediction
    y_pred = w * X

    # Loss
    loss = np.mean((y - y_pred)**2)

    # Gradient
    gradient = -2 * np.mean(X * (y - y_pred))

    # Update rule
    w = w - lr * gradient

    if epoch % 10 == 0:
        print(f'Epoch {epoch}, Loss {loss:.4f}, Weight {w:.4f}')

print('Final Weight:', w)
```

# 🔹 7. Local vs Global Minima

### Global Minimum
- Lowest possible loss

### Local Minimum
- Lowest in a small region

### Why important?
- Linear regression → convex (easy)
- Neural networks → non-convex (hard)

👉 Gradient descent may get stuck in local minima
