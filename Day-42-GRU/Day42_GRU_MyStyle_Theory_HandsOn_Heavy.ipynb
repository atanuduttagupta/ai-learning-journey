# 🧠 Introduction to GRU

GRU (Gated Recurrent Unit) is a type of Recurrent Neural Network (RNN) architecture designed for sequence learning.

It was introduced to solve major problems in traditional RNNs:

❌ Vanishing Gradients  
❌ Forgetting long-term information  
❌ Slow and unstable training  

---

# 🔥 Why GRU Became Popular

GRU introduced:

✅ Better memory handling  
✅ Faster training  
✅ Fewer parameters  
✅ Better gradient flow  
✅ Simpler architecture than LSTM  

---

# 🏗️ GRU Core Idea

Instead of blindly updating memory at every timestep, GRU intelligently decides:

- What to remember
- What to forget
- What to update

using gates.



# ⚠️ Problems with Traditional RNN

## 1️⃣ Vanishing Gradient Problem

During backpropagation:

\[
0.5 \times 0.5 \times 0.5 \times 0.5 \rightarrow 0
\]

Gradients become extremely small.

Result:
- Early layers stop learning
- Long-term memory disappears

---

## 2️⃣ Long-Term Dependency Problem

Example:

> "I was born in France ... I speak fluent French."

Traditional RNN may forget the earlier word "France".

GRU solves this using gated memory control.

---

# 🔄 Difference Between RNN, LSTM, and GRU

| Feature | RNN | LSTM | GRU |
|---|---|---|---|
| Gates | None | 3 | 2 |
| Cell State | No | Yes | No |
| Long-Term Memory | Weak | Strong | Strong |
| Parameters | Low | High | Medium |
| Speed | Fast | Slow | Faster |
| Complexity | Simple | Complex | Simpler |



# 🏛️ GRU Architecture Overview

GRU contains:

✅ Hidden State  
✅ Update Gate  
✅ Reset Gate  
✅ Candidate Hidden State  

---

# 🔐 Update Gate

Controls:

> How much previous memory should be retained?

Equation:

\[
z_t = \sigma(W_z[h_{t-1}, x_t])
\]

If:
- \(z_t \approx 1\) → keep old memory
- \(z_t \approx 0\) → update memory

---

# 🔓 Reset Gate

Controls:

> How much old information should be forgotten?

Equation:

\[
r_t = \sigma(W_r[h_{t-1}, x_t])
\]

---

# 🧪 Candidate Hidden State

Temporary new memory content.

\[
\tilde{h}_t = tanh(W[r_t * h_{t-1}, x_t])
\]

---

# 🧠 Final Hidden State

\[
h_t = (1-z_t)h_{t-1} + z_t\tilde{h}_t
\]



# 🔄 GRU Information Flow

Input → Gates → Candidate Memory → Hidden State Update → Output

The gates dynamically regulate:

✅ Information preservation  
✅ Memory updates  
✅ Forgetting irrelevant context  

---

# 📌 Why GRU Is Efficient

Unlike LSTM:
- No separate cell state
- Fewer gates
- Fewer parameters

Result:
✅ Faster training  
✅ Lower GPU memory usage  
✅ Better computational efficiency  


```python

# 📦 Import Libraries

import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import GRU, Dense, Dropout
from tensorflow.keras.optimizers import Adam
```


# 🛠️ Hands-On Project 1 — Basic GRU Classification

Goal:
Build a simple GRU-based binary classification model.

Dataset:
Synthetic sequence dataset.


```python

# Create Dummy Sequential Data

X = np.random.rand(2000, 30, 8)

y = np.random.randint(0, 2, size=(2000, 1))

print("X Shape:", X.shape)
print("y Shape:", y.shape)
```

```python

# Build GRU Model

model = Sequential([
    
    GRU(
        units=64,
        input_shape=(30, 8),
        return_sequences=False
    ),
    
    Dropout(0.3),
    
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer=Adam(0.001),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python

# Train Model

history = model.fit(
    X,
    y,
    epochs=5,
    batch_size=32,
    validation_split=0.2
)
```


# 📊 Understanding Important GRU Hyperparameters

| Hyperparameter | Meaning |
|---|---|
| units | Hidden neurons |
| return_sequences | Return full sequence or last output |
| dropout | Prevent overfitting |
| batch_size | Samples per training step |
| epochs | Full dataset iterations |
| learning_rate | Optimization step size |

---

# 💡 Practical Insight

Increasing hidden units:
✅ Better learning capacity

# 🧠 Embedding + GRU Workflow

Typical NLP Pipeline:

Text → Tokenization → Embedding → GRU → Dense Layer → Prediction

Embedding converts words into dense vectors.


```python

# Embedding + GRU Example

from tensorflow.keras.layers import Embedding

vocab_size = 10000

model = Sequential([
    
    Embedding(
        input_dim=vocab_size,
        output_dim=64
    ),
    
    GRU(128),
    
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
```


# 🤖 GRU for Sentiment Analysis

Applications:
- Movie review classification

- Social media sentiment mining

Input:
Sequence of words

Output:
Positive / Negative / Neutral


```python

# Sentiment Classification Example

from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.preprocessing.text import Tokenizer

texts = [
    "movie was amazing",
    "bad acting and poor story",
    "excellent performance",
    "terrible film"
]

labels = [1, 0, 1, 0]

tokenizer = Tokenizer(num_words=1000)

tokenizer.fit_on_texts(texts)

sequences = tokenizer.texts_to_sequences(texts)

X = pad_sequences(sequences, maxlen=5)

y = np.array(labels)

print(X)
```


# ⏳ Hands-On Project 2 — Time-Series Forecasting

Goal:
Use GRU for sequence prediction.

Applications:
- Stock prices
- Weather forecasting
- Demand forecasting
- Sensor prediction


```python

# Create Time-Series Data

data = np.sin(np.linspace(0, 100, 3000))

sequence_length = 30

X = []
y = []

for i in range(len(data) - sequence_length):

    X.append(data[i:i+sequence_length])
    y.append(data[i+sequence_length])

X = np.array(X)
y = np.array(y)

X = X.reshape(X.shape[0], X.shape[1], 1)

print(X.shape)
print(y.shape)
```

```python

# GRU Time-Series Model

model = Sequential([
    
    GRU(
        64,
        input_shape=(sequence_length, 1)
    ),
    
    Dense(1)
])

model.compile(
    optimizer='adam',
    loss='mse'
)

model.summary()
```

```python

# Train Time-Series GRU

history = model.fit(
    X,
    y,
    epochs=5,
    batch_size=32
)
```


# 🔄 Bidirectional GRU (BiGRU)

BiGRU processes sequences:
➡️ Forward  
⬅️ Backward  

Benefits:
✅ Better contextual understanding  
✅ Uses future + past context  

Commonly used in:
- NLP
- Translation
- Speech recognition


```python

# Bidirectional GRU Example

from tensorflow.keras.layers import Bidirectional

model = Sequential([
    
    Bidirectional(
        GRU(64)
    ),
    
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
```


# 🏗️ Stacked / Deep GRU

Multiple GRU layers stacked vertically.

Benefits:
✅ Hierarchical feature learning  
✅ Better representation power  

Drawback:
❌ Increased computation


```python

# Stacked GRU Model

model = Sequential([
    
    GRU(
        128,
        return_sequences=True,
        input_shape=(30, 8)
    ),
    
    GRU(64),
    
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

# ⚖️ When to Use GRU Instead of LSTM

Use GRU when:

✅ Faster training needed  
✅ Smaller dataset  
✅ Limited hardware  
✅ Real-time systems  
✅ Lower memory requirements  

Use LSTM when:

✅ Extremely long dependencies exist  
✅ More memory control required  


```python

# 🧪 PyTorch GRU Example

import torch
import torch.nn as nn

class GRUNetwork(nn.Module):

    def __init__(self):

        super().__init__()

        self.gru = nn.GRU(
            input_size=8,
            hidden_size=64,
            batch_first=True
        )

        self.fc = nn.Linear(64, 1)

    def forward(self, x):

        output, hidden = self.gru(x)

        out = self.fc(hidden[-1])

        return out

model = GRUNetwork()

x = torch.randn(32, 30, 8)

output = model(x)

print("Output Shape:", output.shape)
```
