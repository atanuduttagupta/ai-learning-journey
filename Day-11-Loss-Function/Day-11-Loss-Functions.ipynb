# 🧠 What is a Loss Function?

A loss function measures how wrong your model is.

Mathematically:
Loss = f(actual, predicted)

👉 It converts model mistakes into a number that we can minimize.

### Real-world intuition (Self-driving car 🚗)
- Actual steering: 10°
- Predicted: 25°
- Loss = danger level of mistake

👉 Bigger mistake = higher loss

# 📉 Part 1: Regression Loss Functions

## 1️⃣ Mean Squared Error (MSE)

### Formula:
MSE = (1/n) * Σ (y - ŷ)^2

### Intuition:
- Penalizes large errors heavily
- Squaring makes big mistakes dominate

### When to use:
- When large errors are very costly
- Example: Autonomous driving

```python
import numpy as np

# Sample data
y_true = np.array([100, 200, 300])
y_pred = np.array([110, 190, 400])

def mse(y_true, y_pred):
    return np.mean((y_true-y_pred)**2)

print('MSE:', mse(y_true, y_pred))
```

## 2️⃣ Mean Absolute Error (MAE)

### Formula:
MAE = (1/n) * Σ |y - ŷ|

### Intuition:
- Treats all errors equally
- Less sensitive to outliers

### When to use:
- When data has noise/outliers

```python
def mae(y_true, y_pred):
    return np.mean(np.abs(y_true - y_pred))

print('MAE:', mae(y_true, y_pred))
```

## 3️⃣ Huber Loss

### Idea:
- Combines MSE + MAE
- Small errors → squared
- Large errors → absolute

### Why important?
- Robust to outliers
- Stable training

```python
def huber(y_true, y_pred, delta=50):
    error = y_true - y_pred
    condition = np.abs(error) <= delta
    squared_loss = 0.5+ error ** 2
    linear_Loss = delta*(np.abs(error) - .5*delta)
    return np.mean(np.where(condition, squared_loss, linear_Loss))


print('Huber Loss:', huber(y_true, y_pred))
```

# 🔍 Experiment: Effect of Outliers

👉 Add extreme value and compare losses

```python
y_true_out = np.array([100, 200, 300, 10000])
y_pred_out = np.array([110, 190, 400, 0])

print('MSE with outlier:', mse(y_true_out, y_pred_out))
print('MAE with outlier:', mae(y_true_out, y_pred_out))
print('Huber with outlier:', huber(y_true_out, y_pred_out))
```

# 📊 Part 2: Classification Loss Functions

## 4️⃣ Binary Cross Entropy (Log Loss)

### Formula:
L = -[y log(p) + (1-y) log(1-p)]

### Intuition:
- Punishes confident wrong predictions

### Example:
- True = 1, Predicted = 0.9 → Low loss
- True = 1, Predicted = 0.1 → HIGH loss

```python
def log_loss(y_true, y_pred):
    epsilon = 1e-15
    y_pred = np.clip(y_pred, epsilon, 1-epsilon)
    print(y_pred)
    return -np.mean(y_true*np.log(y_pred) + (1-y_true)*np.log(1-y_pred))

y_true_cls = np.array([1, 0, 1])
y_pred_cls = np.array([0.9, 0.2, 0.1])

print('Log Loss:', log_loss(y_true_cls, y_pred_cls))
```

## 5️⃣ Hinge Loss (Used in SVM)

### Formula:
Loss = max(0, 1 - y * f(x))

Where:
- y = +1 or -1
- f(x) = model output

### Intuition:
- Correct classification → low loss
- Wrong side of boundary → penalty

```python
def hinge_loss(y_true, y_pred):
    return np.mean(np.maximum(0, 1 - y_true * y_pred))

y_true_hinge = np.array([1, -1, 1])
y_pred_hinge = np.array([0.8, -0.5, -0.2])

print('Hinge Loss:', hinge_loss(y_true_hinge, y_pred_hinge))
```

# 🧩 Summary

| Loss | Use Case | Behavior |
|------|--------|---------|
| MSE | Regression | Penalizes large errors |
| MAE | Regression | Robust to outliers |
| Huber | Regression | Balanced |
| Log Loss | Classification | Penalizes confidence errors |
| Hinge | Classification | Margin-based learning |
