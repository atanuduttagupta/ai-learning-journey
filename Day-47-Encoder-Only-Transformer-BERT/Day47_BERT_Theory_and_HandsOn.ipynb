# Theory: What is BERT?

**BERT (Bidirectional Encoder Representations from Transformers)** is an Encoder-Only Transformer introduced by Google in 2018.

Key ideas:
1. Bidirectional context understanding
2. Self-Attention mechanism
3. Pretraining + Fine-tuning paradigm

Unlike GPT, BERT reads both left and right context simultaneously.

Example:

The bank is near the river.

BERT understands that 'bank' refers to a river bank because it sees the entire sentence.



# BERT Architecture

Input Text
→ Tokenization
→ Embeddings
→ Positional Encoding
→ Transformer Encoder Layers
→ Contextual Representations

BERT uses only the Encoder stack of the original Transformer.



# Pretraining Tasks


# Load a Pretrained Sentiment Analysis Pipeline


```python

from transformers import pipeline

sentiment_pipeline = pipeline(
    'sentiment-analysis',
    model='distilbert-base-uncased-finetuned-sst-2-english'
)

print("Pipeline loaded successfully")
```


# Test Individual Predictions


```python

texts = [
    "This movie is fantastic",
    "Worst movie ever",
    "Amazing acting and story",
    "Very boring film"
]

results = sentiment_pipeline(texts)

for text, result in zip(texts, results):
    print(f"Text: {text}")
    print(result)
    print("-"*50)
```


# Extract Embeddings from BERT


```python

from transformers import AutoTokenizer, AutoModel
import torch

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
model = AutoModel.from_pretrained("bert-base-uncased")

sentence = "I am learning BERT today"

inputs = tokenizer(sentence, return_tensors="pt")

with torch.no_grad():
    outputs = model(**inputs)

last_hidden_state = outputs.last_hidden_state

print("Shape:", last_hidden_state.shape)
```


# Understanding CLS Token

The first token ([CLS]) is commonly used as the sentence representation for classification tasks.


```python

cls_embedding = last_hidden_state[:,0,:]

print("CLS embedding shape:", cls_embedding.shape)
```

# Summary

- BERT is an Encoder-Only Transformer.
- Uses Bidirectional Self-Attention.
- Pretrained using MLM and NSP.
- Excellent for classification, search, QA, and NER.
- Not designed for text generation.
- Hugging Face makes it easy to use pretrained BERT models.
