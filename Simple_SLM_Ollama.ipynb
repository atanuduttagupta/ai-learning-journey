Writing the Python Implementation
Now that the background server is handling the infrastructure, the Python interaction pipeline becomes incredibly straightforward.

Step 2: Define the Generation Function
Here is the equivalent ask_tinyllama function. Note how the underlying messages schema remains exactly the same as your previous Hugging Face implementation.

```python
import ollama

def ask_tinyllama(prompt):
    # 1. Standardized message schema setup
    messages = [
        {"role": "user", "content": prompt}
    ]

    # 2. Execution / Inference
    # Ollama automatically handles tokenization, device mapping, and detokenization
    response = ollama.chat(
        model='tinyllama',
        messages=messages,
        options={
            'temperature': 0.7,      # Creativity slider
            'num_predict': 150       # Equivalent to max_new_tokens
        }
    )

    # 3. Extract the clean text payload from the structured dictionary response
    return response['message']['content']
```

Step 3: Run an Execution Test

```python
response = ask_tinyllama("Explain what an API is in one simple sentence.")
print(response)
```
