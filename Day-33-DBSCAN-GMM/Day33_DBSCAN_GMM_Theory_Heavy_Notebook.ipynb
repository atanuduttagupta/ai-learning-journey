# 1. Introduction to Advanced Clustering

Traditional clustering methods like K-Means assume:
- Clusters are spherical
- Similar density
- Minimal noise

Real-world datasets often violate these assumptions.

Advanced clustering methods:
- DBSCAN → Density-Based Clustering
- GMM → Distribution-Based Clustering

---

# 2. What is DBSCAN?

DBSCAN stands for:

## Density-Based Spatial Clustering of Applications with Noise

Core idea:
- Dense regions form clusters
- Sparse regions separate clusters
- Isolated points become noise

---

# 3. Important Concepts in DBSCAN

## Epsilon (eps)
Neighborhood radius around a point.

## MinPts
Minimum neighbors needed to form dense region.

## Core Point
Point having at least MinPts neighbors.

## Border Point
Point connected to core point but not dense enough itself.

## Noise Point
Outlier not belonging to any cluster.

---

# 4. How DBSCAN Works

1. Pick random point
2. Find neighbors within eps
3. If neighbors >= MinPts:
   - Create cluster
4. Expand cluster using connected dense points
5. Repeat until all points processed

---

# 5. DBSCAN Python Implementation


```
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons
import matplotlib.pyplot as plt

# Generate non-linear data
X, y = make_moons(n_samples=300, noise=0.05, random_state=42)

# Train DBSCAN model
dbscan = DBSCAN(eps=0.2, min_samples=5)

clusters = dbscan.fit_predict(X)

# Plot clusters
plt.figure(figsize=(8,6))
plt.scatter(X[:,0], X[:,1], c=clusters)
plt.title("DBSCAN Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

# 9. What is GMM?

Gaussian Mixture Model assumes:
- Data generated from multiple Gaussian distributions
- Each cluster has probability distribution

This is called:

# Soft Clustering

---

# 10. Hard vs Soft Clustering

| Type | Meaning |
|---|---|
| Hard Clustering | One point belongs to one cluster |
| Soft Clustering | Probabilistic membership |

Example:
- 70% Cluster A
- 30% Cluster B

---

# 11. Gaussian Distribution

Normal distribution:
- Bell-shaped curve
- Defined by:
  - Mean
  - Variance

---

# 12. EM Algorithm in GMM

## Step 1: Initialize
Random Gaussian distributions

## Step 2: E-Step
Compute probabilities

## Step 3: M-Step
Update:
- Mean
- Variance
- Mixing weights

## Step 4:
Repeat until convergence

---

# 13. GMM Python Implementation


```
from sklearn.mixture import GaussianMixture
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt

# Generate sample data
X, y = make_blobs(n_samples=300, centers=3, random_state=42)

# Train GMM
gmm = GaussianMixture(n_components=3, random_state=42)

gmm.fit(X)

clusters = gmm.predict(X)

# Plot
plt.figure(figsize=(8,6))
plt.scatter(X[:,0], X[:,1], c=clusters)
plt.title("Gaussian Mixture Model Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```
