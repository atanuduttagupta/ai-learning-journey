# Day 02 - Cosine Similarity

Cosine similarity is widely used in AI systems such as:
- Semantic search
- Recommendation systems
- NLP embeddings
- RAG (Retrieval Augmented Generation)

**It helps measure how similar two vectors are based on their direction.**

How do we measure similarity between two vectors?

Instead of comparing values directly, cosine similarity measures the **angle between vectors**.

- Smaller angle → more similar  
- Larger angle → less similar

***Data***

```python
v1 = [1, 2, 3]
v2 = [5, -1, 7]
```

***Approach 1: Traditional (Manual code for Understanding the derivation logic)***

Steps:
1. Compute dot product
2. Compute magnitude of each vector
3. Apply cosine similarity formula

```python
import math

def cosine_similarity_traditional(v1, v2):
    dot_product = sum(a * b for a, b in zip(v1, v2))
    magnitude_v1 = math.sqrt(sum(a * a for a in v1))
    magnitude_v2 = math.sqrt(sum(b * b for b in v2))
    return dot_product / (magnitude_v1 * magnitude_v2)

cosine_similarity_traditional(v1, v2)
```

***Approach 2: Pythonic (NumPy)***

Using NumPy makes the code:
- Cleaner
- Faster
- More suitable for large-scale AI applications

```python
import numpy as np

def cosine_similarity_pythonic(v1, v2):
    v1 = np.array(v1)
    v2 = np.array(v2)
    cosine_similarity = np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))
    return float(cosine_similarity)

cosine_similarity_pythonic(v1, v2)
```

***Comments***

Cosine similarity depends on the **direction of vectors**, not their magnitude. It is used widely in

- NLP → comparing sentence embeddings  
- Search → finding similar documents  
- Recommendation systems → matching user preferences  
- RAG pipelines → retrieving relevant context
