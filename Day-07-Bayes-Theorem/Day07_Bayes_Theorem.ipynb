# Day 07 - Bayes Theorem (Medical Diagnosis)

### Key Question:
**If a test is 99% accurate and you test positive, do you really have the disease?**


```python
# Step 1: Define probabilities
P_disease = 0.01
P_no_disease = 0.99

P_pos_given_disease = 0.99
P_pos_given_no_disease = 0.01

print("Setup complete")
```

```python
# Step 2: Total probability of positive
P_positive = (P_pos_given_disease * P_disease) + (P_pos_given_no_disease * P_no_disease)
print("P(Positive):", round(P_positive, 4))
```

```python
# Step 3: Bayes Theorem
P_disease_given_positive = (P_pos_given_disease * P_disease) / P_positive
print("P(Disease | Positive):", round(P_disease_given_positive, 4))
```

```python
# Step 4: 1000 people intuition
population = 1000
disease_people = population * P_disease
healthy_people = population * P_no_disease

true_positives = disease_people * P_pos_given_disease
false_positives = healthy_people * P_pos_given_no_disease

print("True Positives:", int(true_positives))
print("False Positives:", int(false_positives))
```

```python
#step 5: Change disease probability
for prevalence in [0.01, 0.05, 0.1, 0.2]:
    P_disease = prevalence
    P_no_disease = 1 - prevalence

    P_positive = (P_pos_given_disease * P_disease) + \
                 (P_pos_given_no_disease * P_no_disease)

    P_disease_given_positive = (P_pos_given_disease * P_disease) / P_positive

    print(f"Disease Rate: {prevalence*100:.0f}% -> Probability after test: {round(P_disease_given_positive, 3)}")
```
