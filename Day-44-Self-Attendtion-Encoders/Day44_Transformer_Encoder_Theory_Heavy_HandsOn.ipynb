# Word Embedding

## Theory

Transformers cannot process text directly.

Words are converted into dense vectors called embeddings.

### Why Embeddings?

- Neural networks understand numbers.
- Similar words have similar vectors.
- Capture semantic relationships.

Example:
King − Man + Woman ≈ Queen


## Hands-On Exercise

```python

words = ["I", "love", "AI"]
embeddings = {
    "I": [0.12, -0.45, 0.87],
    "love": [0.31, 0.92, -0.11],
    "AI": [-0.76, 0.23, 0.45]
}

for word in words:
    print(word, "→", embeddings[word])
```

# Positional Encoding

## Theory

Transformers process all words simultaneously.

Without positional encoding:

Dog bites man
Man bites dog

would appear identical.

### Original Formula

PE(pos,2i)=sin(...)
PE(pos,2i+1)=cos(...)


## Hands-On Exercise

```python

import numpy as np

def positional_encoding(position, d_model):
    pe = np.zeros((position, d_model))

    for pos in range(position):
        for i in range(0, d_model, 2):
            pe[pos, i] = np.sin(pos / (10000 ** (i / d_model)))
            if i + 1 < d_model:
                pe[pos, i+1] = np.cos(pos / (10000 ** (i / d_model)))

    return pe

pe = positional_encoding(5, 6)
print(pe)
```

# Self-Attention

## Theory

Self-Attention asks:

Which words should I focus on?

### QKV

- Query: What am I looking for?
- Key: What do I contain?
- Value: What information should I provide?

### Formula

Attention(Q,K,V)
= softmax(QKᵀ/√dk)V


## Hands-On Exercise

```python

import numpy as np

Q = np.array([[1,0,1]])
K = np.array([[1,1,0],[0,1,1]])
V = np.array([[1,2],[3,4]])

scores = Q @ K.T
weights = np.exp(scores) / np.sum(np.exp(scores))

output = weights @ V

print("Scores:", scores)
print("Weights:", weights)
print("Attention Output:", output)
```

# Residual Connection

## Theory

Deep networks may forget earlier information.

Residual Connection:

Output = Input + Sublayer(Input)

Benefits:
- Better gradient flow
- Stable training
- Preserves original information


## Hands-On Exercise

```python

import numpy as np

x = np.array([1,2,3])
attention_output = np.array([0.5,0.2,0.8])

residual = x + attention_output

print("Input:", x)
print("Attention:", attention_output)
print("Residual Output:", residual)
```
