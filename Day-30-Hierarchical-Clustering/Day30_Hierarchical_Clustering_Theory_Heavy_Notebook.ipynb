# 🌳 What is Hierarchical Clustering?

Hierarchical Clustering is an **unsupervised machine learning algorithm** used to group similar data points into clusters.

Unlike KMeans:
- It does NOT require pre-defining number of clusters initially.
- It creates a hierarchy (tree-like structure) of clusters.
- Output is visualized using a **Dendrogram**.

The algorithm gradually:
- merges clusters → Agglomerative
OR
- splits clusters → Divisive


# 🌳 What is a Dendrogram?

A dendrogram is a tree-like diagram showing:
- how clusters merge
- similarity between clusters
- distance between data points

It helps determine:
- optimal number of clusters
- cluster relationships
- cluster hierarchy

## Key Interpretation
- Bottom nodes = individual data points
- Vertical lines = cluster merging
- Height = cluster distance/similarity

Higher merge height = lower similarity.


# 📌 Types of Hierarchical Clustering

| Type | Process |
|---|---|
| Agglomerative | Bottom → Top (merge clusters) |
| Divisive | Top → Bottom (split clusters) |


# 1️⃣ Agglomerative Clustering

## Definition
Agglomerative means:

> “Start small and merge gradually.”

Initially:
- every data point is its own cluster.

Then:
- nearest clusters are merged repeatedly.

Until:
- all points become one cluster
OR
- desired number of clusters is achieved.

## Process
1. Each point becomes its own cluster.
2. Compute distances.
3. Merge closest clusters.
4. Recompute distances.
5. Repeat.


# 2️⃣ Divisive Clustering

## Definition
Divisive means:

> “Start big and divide gradually.”

Initially:
- all points belong to one cluster.

Then:
- recursively split clusters.

Until:
- desired clusters obtained.

## Important
Agglomerative clustering is far more common in industry because divisive clustering is computationally expensive.


# 🔗 Linkage Methods

A major question:

> “How do we measure distance between clusters?”

This is solved using linkage methods.


# 📌 Single Linkage

## Definition
Distance between closest points of two clusters.

## Advantages
- Detects irregular cluster shapes
- Good for connected structures

## Limitations
- Sensitive to noise
- Chaining effect
- Creates stretched clusters


# 📌 Complete Linkage

## Definition
Distance between farthest points of two clusters.

## Advantages
- Produces compact clusters
- Less chaining effect

## Limitations
- Sensitive to outliers
- Can break natural clusters


# 📌 Average Linkage

## Definition
Average distance between all points of two clusters.

## Advantages
- Balanced approach
- Stable clustering

## Limitations
- Computationally expensive
- Smooths boundaries


# 📌 Ward Linkage

## Definition
Merges clusters causing minimum increase in variance.

## Core Idea
Ward linkage minimizes:
- cluster variance
- cluster spread
- within-cluster SSE

Very similar philosophy to KMeans.

## Advantages
- Compact clusters
- Very popular in industry

## Limitations
- Sensitive to outliers
- Prefers spherical clusters


# 📏 Distance Metrics

| Metric | Usage |
|---|---|
| Euclidean | Most common |
| Manhattan | Grid distance |
| Cosine | Text similarity |
| Minkowski | Generalized distance |


# 🧮 Euclidean Distance Formula

\[
d = \sqrt{(x_2-x_1)^2 + (y_2-y_1)^2}
\]

Measures straight-line distance between points.

4. Choose linkage method
5. Build dendrogram
6. Cut dendrogram
7. Form clusters

# ✅ Advantages of Hierarchical Clustering

1. No need to predefine K
2. Dendrogram visualization
3. Detects nested clusters
4. Flexible distance metrics
5. Excellent for small datasets


# ❌ Limitations of Hierarchical Clustering

1. Computationally expensive
2. Memory intensive
3. Sensitive to noise
4. Greedy process
5. Poor scalability


# 🔥 Hierarchical Clustering vs KMeans

| Feature | Hierarchical | KMeans |
|---|---|---|
| Type | Tree-based | Centroid-based |
| Need K Initially | No | Yes |
| Output | Dendrogram | Clusters |
| Scalability | Poor | Excellent |
| Speed | Slow | Fast |
| Interpretability | High | Medium |
| Handles Complex Shapes | Better | Weak |


# 🧠 Important Interview Questions

## ❓ Why use dendrogram?
To visually determine optimal clusters.

## ❓ Why is hierarchical clustering expensive?
Because pairwise distances are repeatedly computed.

## ❓ Why is Ward linkage popular?
Because it minimizes variance and creates compact clusters.

## ❓ Difference between Agglomerative and Divisive?
- Agglomerative → merge clusters
- Divisive → split clusters


# 🚀 Key Takeaways

✅ Hierarchical clustering builds cluster hierarchy.

✅ Dendrogram visualizes cluster merging.

✅ Agglomerative clustering is most commonly used.

✅ Ward linkage often produces best compact clusters.

✅ Better interpretability than KMeans.

# 💻 Practical Agglomerative Clustering Example

```python
from sklearn.cluster import AgglomerativeClustering
from sklearn.datasets import make_blobs
from scipy.cluster.hierarchy import dendrogram, linkage
import matplotlib.pyplot as plt

# Generate data
X, y = make_blobs(
    n_samples=30,
    centers=4,
    random_state=42
)

# -------------------------------
# DENDROGRAM
# -------------------------------

linked = linkage(X, method='ward')

plt.figure(figsize=(14,6))

dendrogram(
    linked,
    leaf_rotation=90
)

plt.title("Dendrogram")
plt.xlabel("Data Points")
plt.ylabel("Distance")

plt.show()

# -------------------------------
# AGGLOMERATIVE CLUSTERING
# -------------------------------

model = AgglomerativeClustering(
    n_clusters=4,
    linkage='ward'
)

labels = model.fit_predict(X)

plt.figure(figsize=(8,6))

plt.scatter(
    X[:,0],
    X[:,1],
    c=labels
)

plt.title("Agglomerative Clustering")

plt.show()
```
