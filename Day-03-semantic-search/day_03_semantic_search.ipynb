# Day 03 - Semantic Search

## 🧠 Why this matters

Traditional search looks for exact word matches.

But modern AI systems understand **meaning**, not just words.

This is called **semantic search**.

## 📌 Problem

What happens when users use different words for the same idea?

Example:
- "couch" vs "sofa"

Traditional systems fail here.

```python
text = "I bought a new sofa"
query = "couch"
if query in text:
    print("Match")
else:
    print("No Match")
```

## ⚠️ Limitation

Even though "couch" and "sofa" mean the same thing,
keyword search fails because it only matches exact words.

## 🔹 Semantic Search (AI Approach)

Instead of comparing words directly:

1. Convert sentences into numbers (vectors)
2. Compare their similarity
3. Return the closest match

This helps AI understand meaning.

```python
!pip install sentence-transformers
```

```python
from sentence_transformers import SentenceTransformer, util
from transformers import logging
logging.set_verbosity_error()

model = SentenceTransformer('all-MiniLM-L6-v2')

s1 = "The weather is freezing"
s2 = "It is very cold outside"

v1 = model.encode(s1)
v2 = model.encode(s2)

similarity = util.cos_sim(v1, v2)

print('Similarity Score: ', similarity.item())
```

## 📊 Observation

Different words… same meaning.

👉 AI captures similarity using vectors.

```python
query = "I love AI"
documents = [
    "I enjoy machine learning",
    "The sky is nice",
    "Artificial Intelligence rocks"
]

doc_embeddings = model.encode(documents)
query_embedding = model.encode(query)

scores = util.cos_sim(query_embedding, doc_embeddings)

best_index = scores[0].argmax()
best_match = documents[best_index]
score = float(scores[0][best_index])

print("Semantic Search:")
print(f"Query: {query}")
print(f"Best Match: {best_match}")
print(f"Score: {score:.4f}")
```

## 🚀 AI Connection

Semantic search is used in:

- Google Search
- ChatGPT / LLMs
- Recommendation systems
- Vector databases

This is a core building block of modern AI systems.
