# Transformer Family

```text
Transformer
│
├── Encoder Only
│      └── BERT
│
├── Decoder Only
│      └── GPT
│
└── Encoder-Decoder
       ├── T5
       └── BART
```

Encoder = Understand

Decoder = Generate

Encoder-Decoder = Understand + Generate



# Why Encoder-Decoder?

Many tasks require converting one sequence into another.

Examples:

- Translation
- Summarization
- Question Answering
- Paraphrasing

Input and output are different sequences.

This is called Sequence-to-Sequence (Seq2Seq) Learning.



# Cross-Attention

Encoder creates contextual representations.

Decoder uses:

1. Masked Self-Attention
2. Cross-Attention

Cross-Attention allows the decoder to look at encoder outputs while generating text.

This is the key innovation behind translation and summarization.



# T5 (Text-To-Text Transfer Transformer)

Developed by Google.

Core Idea:

Everything becomes:

Text ➜ Text

Examples:

translate English to French:
I love AI

summarize:
<long article>

question:
Where is Taj Mahal?



# BART (Bidirectional and Auto-Regressive Transformer)

Developed by Meta.

Think of BART as:

BERT + GPT

Encoder:
- Bidirectional

Decoder:
- Autoregressive

Strong at:
- Summarization
- Generation
- Document reconstruction


# 🧪 Hands-On 1: Translation with T5


````python

import torch
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
````

````python
# 🧪 Hands-On 1: Translation using FLAN-T5
````

````python
model_name = "google/flan-t5-small"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

text = "Translate English to French: I love artificial intelligence"

inputs = tokenizer(text, return_tensors="pt")

outputs = model.generate(
    **inputs,
    max_new_tokens=50
)

result = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print(result)
````

#🧪 Hands-On 2: Question Answering

````python
question = """
Answer the question.

Question: Where is the Taj Mahal?

Context: The Taj Mahal is located in Agra, India.
"""

inputs = tokenizer(
    question,
    return_tensors="pt"
)

outputs = model.generate(
    **inputs,
    max_new_tokens=20
)

answer = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print(answer)
````

# Hands-On 3: Summarization using FLAN-T5

````python
article = """
Artificial Intelligence is transforming industries worldwide.
Companies are investing heavily in AI technologies to improve
efficiency, automate repetitive tasks, and gain business insights.
Machine learning and deep learning are becoming essential tools
for modern enterprises.
"""

prompt = f"Summarize this text:\n\n{article}"

inputs = tokenizer(
    prompt,
    return_tensors="pt",
    truncation=True
)

outputs = model.generate(
    **inputs,
    max_new_tokens=60
)

summary = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print(summary)
````

# Hands-On 4: Paraphrasing

````python
text = """
Paraphrase:

Artificial Intelligence is changing the world.
"""

inputs = tokenizer(
    text,
    return_tensors="pt"
)

outputs = model.generate(
    **inputs,
    max_new_tokens=60
)

result = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print(result)
````

# Hands-On 5: Multiple Language Translation

````python
languages = [
    "French",
    "German",
    "Spanish",
    "Hindi"
]

for lang in languages:

    prompt = f"""
    Translate English to {lang}:
    Artificial Intelligence is amazing.
    """

    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    )

    outputs = model.generate(
        **inputs,
        max_new_tokens=60
    )

    translation = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    print(f"{lang}: {translation}")
````


# 🧪 Hands-On 6: BART Summarization


````python
from transformers import BartTokenizer, BartForConditionalGeneration
````

````python
model_name = "facebook/bart-large-cnn"

tokenizer = BartTokenizer.from_pretrained(model_name)
model = BartForConditionalGeneration.from_pretrained(model_name)
````

````python
article = """
Artificial Intelligence is transforming industries worldwide.
Organizations are using AI to automate operations,
improve customer experience, and make data-driven decisions.kjggukkuvvjb about a tiger and dog.
dog is our frirnd and not a friend of ai
"""

inputs = tokenizer(
    article,
    return_tensors="pt",
    max_length=1024,
    truncation=True
)

summary_ids = model.generate(
    inputs["input_ids"],
    max_length=60,
    min_length=10,
    num_beams=4
)

summary = tokenizer.decode(
    summary_ids[0],
    skip_special_tokens=True
)

print(summary)
````

# Hands-On 7: BFSI Complaint Summarizer

````python
complaint = """
I have been trying to contact customer support for three weeks.
My transaction failed but the money was debited from my account.
Multiple complaints were raised but no resolution has been provided.
I am so so sad
I have stopped eating
I am crying all day
I need lot of help
God will bless you
please help
"""

inputs = tokenizer(
    complaint,
    return_tensors="pt",
    truncation=True
)

summary_ids = model.generate(
    inputs["input_ids"],
    max_length=40,
    min_length=10
)

summary = tokenizer.decode(
    summary_ids[0],
    skip_special_tokens=True
)

print(summary)
````
