# 1️⃣ What is a Loss Function?

A **Loss Function** measures how wrong a model's prediction is compared to the actual answer.

Think of it as:

🧠 **"Mistake Score of the Neural Network"**

---

# Example

Suppose:

Actual House Price = ₹50 Lakhs  
Predicted Price = ₹60 Lakhs

The model made a mistake.

Loss Function calculates:
- how big the mistake is
- how badly the model should be punished

---

# Why Loss Functions Matter

Without loss functions:
- neural networks cannot learn
- optimizers cannot update weights
- backpropagation cannot work

The loss function gives the model:
- learning direction
- correction signal
- optimization target

---

# Regression

Used when output is a number.

---

# Classification

# 3️⃣ Mean Squared Error (MSE)

## Formula

\[
MSE = \frac{1}{n}\sum (y - \hat{y})^2
\]

---

# Intuition

- Find prediction error
- Square the error
- Take average

Large mistakes become MUCH larger because of squaring.

---

# Advantages

✔️ Smooth gradients  
✔️ Strongly punishes large errors  
✔️ Works well in regression  

---

# Limitations

❌ Sensitive to outliers  
❌ One huge mistake can dominate training  

---

# Practical Uses

- Price prediction
- Forecasting
- Numerical prediction systems


```python

import numpy as np

actual = np.array([50, 60, 70, 80])
predicted = np.array([52, 65, 68, 78])

mse = np.mean((actual - predicted) ** 2)

print("Actual Values :", actual)
print("Predicted Values :", predicted)
print("MSE :", mse)
```


# 4️⃣ Mean Absolute Error (MAE)

## Formula

\[
MAE = \frac{1}{n}\sum |y - \hat{y}|
\]

---

# Intuition

Instead of squaring:
- we take absolute difference

Every mistake contributes proportionally.

---

# Advantages

✔️ Robust against outliers  
✔️ Easy interpretation  

---

# Limitations

❌ Gradients are less smooth  
❌ Can converge slower than MSE  

---

# Real Use

Used when:
- dataset contains noise
- extreme values should not dominate


```python

import numpy as np

actual = np.array([50, 60, 70, 80])
predicted = np.array([52, 65, 68, 78])

mae = np.mean(np.abs(actual - predicted))

print("MAE :", mae)
```


# 5️⃣ Cross Entropy — The Most Important DL Loss 🔥

Cross Entropy is used in:
- CNNs
- Transformers
- GPT
- BERT
- Image Classification
- NLP
- Chatbots

---

# Core Intuition

Cross Entropy punishes:
- wrong predictions
- especially confidently wrong predictions

---

# Cat vs Dog Example

Actual:

Cat = 1  
Dog = 0

---

## Good Prediction

Cat = 0.95  
Dog = 0.05

Very small loss ✅

---

## Bad Prediction

Cat = 0.10  
Dog = 0.90

Large loss ❌

---

## Confidently Wrong

Cat = 0.01  
Dog = 0.99

Huge penalty 🚨

---

# Why?

Because DL wants the model to be:
- confidently correct
- not confidently wrong



# 6️⃣ Understanding log() in Cross Entropy

Cross Entropy uses logarithm.

For Binary Cross Entropy:

\[
Loss = -log(predicted\ probability)
\]

---

# Example 1

\[
-log(0.95)
\]

\[
= 0.051
\]

Tiny loss ✅

---

# Example 2

\[
-log(0.1)
\]

\[
= 2.302
\]

Large loss ❌

---

# Example 3

\[
-log(0.01)
\]

\[
= 4.605
\]

Huge loss 🚨

---

# Important Observation

As probability for correct answer decreases:
- loss increases rapidly

This is the magic of Cross Entropy.


```python

import numpy as np

good_prediction = 0.95
bad_prediction = 0.10
very_bad_prediction = 0.01

print("Good Prediction Loss :", -np.log(good_prediction))
print("Bad Prediction Loss :", -np.log(bad_prediction))
print("Very Bad Prediction Loss :", -np.log(very_bad_prediction))
```


# 7️⃣ Training Loss vs Validation Loss

| Metric | Meaning |
|---|---|
| Training Loss | Error on training data |
| Validation Loss | Error on unseen data |

---

# Interpretation

## Healthy Learning
- Training Loss ↓
- Validation Loss ↓

✅ Good Generalization

---

## Overfitting

- Training Loss ↓
- Validation Loss ↑

❌ Model memorizing training data

---

## Underfitting

Both losses remain high.

❌ Model not learning properly

# 9️⃣ Mini Project — Simple Loss Function Explorer 🚀

## Goal

Build a tiny simulator that:
- accepts prediction probabilities
- computes Cross Entropy loss
- shows how confidence affects punishment

---

# Expected Learning

By doing this mini project you will understand:
- probability learning
- log behavior
- why confident wrong predictions are dangerous


```python

import numpy as np

while True:
    p = float(input("Enter probability for correct class (0 to 1): "))

    if p <= 0 or p >= 1:
        print("Please enter value between 0 and 1")
        continue

    loss = -np.log(p)

    print(f"Cross Entropy Loss : {loss:.4f}")
    print("-" * 40)
```
