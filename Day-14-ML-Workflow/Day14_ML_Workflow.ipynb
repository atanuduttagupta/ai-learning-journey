## 🧪 Mini Project: Credit Risk Prediction (End-to-End)

---

### Step 1: Generate Data


```python
import numpy as np

np.random.seed(42)

income = np.random.randint(20000, 100000, 100)
loan = np.random.randint(5000, 50000, 100)

risk = (loan / income) + np.random.normal(0, 0.05, 100)
```

### Step 2: Feature Engineering

```python
X = np.column_stack((income, loan, loan/income))
y = risk
```

### Step 3: Train Model

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X, y)
```

### Step 4: Prediction

```python
model.predict([[500000, 20000, 20000/50000]])
```
