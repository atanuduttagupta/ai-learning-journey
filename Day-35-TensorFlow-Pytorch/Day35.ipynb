````python
# Import PyTorch
import os
os.environ["CUDA_VISIBLE_DEVICES"] = ""
os.environ["KMP_DUPLICATE_LIB_OK"] = "TRUE"
import torch

#import torch

print("PyTorch Version:", torch.__version__)
````

````python
# Check GPU availability
print("CUDA Available:", torch.cuda.is_available())
````

````python

# This sets 'cpu' as your default device since CUDA is not present
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# Example: Creating a tensor and sending it to your available device
x = torch.rand(3, 3).to(device)
print(x)
````

````python
# Scalar Tensor
scalar = torch.tensor(5)
print(scalar)

# 1D Tensor
vector = torch.tensor([1,2,3])
print(vector)

# 2D Tensor
matrix = torch.tensor([[1,2],[3,4]])
print(matrix)

# 3D Tensor
matrix = torch.tensor([[[1,2],[3,4]],[[10,2],[30,4]]])
print(matrix)

print("Shape:", matrix.shape)
print("Datatype:", matrix.dtype)
print("Dimensions:", matrix.ndim)
````

````python
# Creating tensors

zeros = torch.zeros((2,3))
ones = torch.ones((2,3))
random_tensor = torch.rand((2,3))

print("Zeros:\n", zeros)
print("Ones:\n", ones)
print("Random Tensor:\n", random_tensor)
````

````python
# Tensor shape and datatype

x = torch.tensor([[1,2,3],[4,5,6]])

print("Shape:", x.shape)
print("Datatype:", x.dtype)
print("Dimensions:", x.ndim)
````

````python
# Arithmetic operations

a = torch.tensor([1,2,3])
b = torch.tensor([4,5,6])

print("Addition:", a+b)
print("Subtraction:", a-b)
print("Multiplication:", a*b)
print("Division:", a/b)
````

````python
# Matrix multiplication

matrix1 = torch.tensor([[1,2],[3,4]])
matrix2 = torch.tensor([[5,6],[7,8]])

result = torch.matmul(matrix1, matrix2)

print(result)
````

````python
# Statistical operations

x = torch.tensor([1,2,3,4,5], dtype=torch.float32)

print("Mean:", torch.mean(x))
print("Sum:", torch.sum(x))
print("Max:", torch.max(x))
print("Min:", torch.min(x))
````

````python
# Indexing and slicing

x = torch.tensor([[1,2,3],[4,5,6]])

print("First Row:", x[0])
print("Second Column:", x[:,1])
````

````python
# Conditional selection

x = torch.tensor([1,2,3,4,5,6])

print(x[x > 3])
````

````python
# Reshaping tensors

x = torch.arange(12)

reshaped = x.reshape(3,4)

print(reshaped)
````

````python
# Unsqueeze and squeeze

x = torch.tensor([1,2,3])

print("Original Shape:", x.shape)

x_unsqueezed = x.unsqueeze(0)

print("After Unsqueeze:", x_unsqueezed.shape)

x_squeezed = x_unsqueezed.squeeze()

print("After Squeeze:", x_squeezed.shape)
````

````python
# Broadcasting example

a = torch.tensor([[1],[2],[3]])
b = torch.tensor([10,20,30])

print(a + b)
````

````python
# NumPy to Tensor

import numpy as np

arr = np.array([1,2,3])

tensor = torch.from_numpy(arr)

print(tensor)
````

````python
# Tensor to NumPy

tensor = torch.tensor([4,5,6])

array = tensor.numpy()


print(array)
````

````python
# Moving tensor to GPU

device = "cuda" if torch.cuda.is_available() else "cpu"

x = torch.tensor([1,2,3]).to(device)

print(x)
print("Device:", x.device)
````

````python
# Autograd example

x = torch.tensor(2.0, requires_grad=True)

y = x**2

y.backward()

print(x)

print("Gradient:", x.grad)
````

````python
# More autograd operations

x = torch.tensor(3.0, requires_grad=True)

y = 3*x**2 + 2*x + 1

y.backward()

print("dy/dx =", x.grad)
````

````python
# Mini Tensor Calculator

a = torch.tensor(10.0)
b = torch.tensor(5.0)

print("Addition:", a+b)
print("Subtraction:", a-b)
print("Multiplication:", a*b)
print("Division:", a/b)
````
